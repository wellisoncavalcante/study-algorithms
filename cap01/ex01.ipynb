01) Suponha que você tenha uma lista com 128 nomes e esteja fazendo uma pesquisa binária. Qual seria o número máximo de etapas que você levaria para encontrar o nome desejado?

R = 7 etapas. 2^7 = 128


02) Suponha que você duplique o tamanho da lista. Qual seria máximo de etapas agora?

R = 8 etapas. 2^8 = 256
