O problema do caixeiro viajante é um problema de O(n!). Um problema muito mas muito lento!

Existem 5 cidades
O caixeiro deseja passar por todas as cidades percorrendo uma distância mínima. Ele soma a distância total e escolhe o caminho de menor distância. Existem 120 permutações para cinco cidades, logo, precisa-se de 120 operações para resolver o problema de cinco cidades.

Para 6 cidades, seriam necessários 720 operações.

Para 7 cidades, seriam necessários 5050 operações

O número de operações aumenta drasticamente.
