Você tem um nome e deseja encontrar o número de telefone para esse nome em uma agenda telefônica.

Se a agenda telefônica estiver ordenada por ordem alfabética pelos nomes, podemos utilizar a pesquisa binária O(log n).
Senão, só poderiamos utilizar a pesquisa linear O(n).

Você tem um número de telefone e deseja encontrar o dono dele em uma agenda telefônica. 
(Dica: Deve procurar pela agenda inteira!)

Então deve-se utilizar a pesquisa linear que é O(n) para fazer uma busca na agenda inteira.
