# 01. Você tem um nome e deseja encontrar o número de telefone para esse nome em uma agenda telefônica.

## Minha solução:
Se a agenda telefônica estiver ordenada por ordem alfabética pelos nomes, podemos utilizar a pesquisa binária O(log n).
Senão, só poderiamos utilizar a pesquisa linear O(n).

# 02. Você tem um número de telefone e deseja encontrar o dono dele em uma agenda telefônica. (Dica: Deve procurar pela agenda inteira!)

## Minha solução: 
Então deve-se utilizar a pesquisa linear que é O(n) para fazer uma busca na agenda inteira.

# 03. Você quer ler o número de cada pessoa da agenda telefônica.
## Minha solução: 
Então deve-se utilizar a pesquisa linear que é O(n) para ler o número de cada pessoa da agenda telefônica.

# 04. Você quer ler os números apenas dos nomes que começam com A.
## Minha solução: 
Mesmo que apenas uma parte dos nomes comece com a letra A, essa quantidade pode crescer proporcionalmente ao tamanho da agenda. Se aproximadamente 1/26 dos nomes começam com A, seriam realizadas n / 26 leituras.

O (n / 26) = O(n)
Portanto, um algoritmo O(n).
