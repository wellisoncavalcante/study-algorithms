# Questão 01.

Escreva o código para a função soma, vista anteriormente.




```python
def soma(lista):
    if lista == []:
        return 0
    else:
        return lista[0] + soma(lista[1:])

print(soma([2, 4, 6]))
```

# Questão 3:
Encontre o valor mais alto em uma lista.

```python
def lista(x):
    return max(x)
print(lista([1, 2, 3, 4, 5]))
```
