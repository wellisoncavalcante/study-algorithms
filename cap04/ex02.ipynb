Quanto tempo levaria, em notação Big O, para completar cada uma destas operações?

# Questão 1.
Imprimir o valor de cada elemento em um array.

## Minha solução:
> O(n)

# Questão 2.
Duplicar o valor de cada elemento em um array.

## Minha solução:
> O(n)

# Questão 3.
Duplicar o valor apenas do primeiro elemento do array.

## Minha solução:
> O(1)

# Questão 4.
Criar uma tabela de multiplicação com todos os elementos do array. Assim, caso o seu array seja [2, 3, 7, 8, 10], você primeiro multiplicará cada elemento por 2. Depois, multiplicará cada elemento por 3 e então por 7, e assim por dianate.

## Minha solução:
> O(n^2).
> Porque irá percorrer todos os elementos para cada quantidade de elemento.
