# Questão 01.
Suponha que eu forneça uma pilha de chamada como esta:



Quais informações você pode retirar baseando-se apenas nesta pilha de chamada?

## Minha solução:

A função **sauda** foi chamada primeiro com o argumento "Maggie" e, durante sua execução, chamou a função **sauda2**, também com "Maggie". Como **sauda2** está no topo da pilha, ela está sendo executada atualmente, enquanto **sauda** está pausada aguardando seu término. Quando **sauda2** finalizar, ela será retilhada da pilha e **sauda** continuará sua execução.

> A última função que entra na pilha é a primeira que sai!

# Questão 02.
Suponha que você acidentalmente escreva uma função recursiva que fique executando infinitamente. Como você viu, seu computador aloca memória na pilha para cada chamada de função. O que acontece com a pilha quando a função recursiva fica executando infinitamente?

## Minha solução:

> A pilha continuará recebendo novas chamadas de função até atingir o seu limite de memória, causando um erro de estouro da pilha.
