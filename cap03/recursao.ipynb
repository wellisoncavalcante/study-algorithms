A recursão é uma técnica de programação utilizada em muitos algoritmos. Você irá aprender como separar um problema em caso-base e caso recursivo. A estratégia dividir para conquistar usa este conceito simples para resolver problemas complicados.

Suponha que você esteja vasculhando o porão de sua avó e encontre uma misteriosa mala trancada.



A sua avó diz que a chave para a mala provavelmente está em uma caixa



Esta caixa contém mais caixas com mais caixas dentro delas. A chave está em alguma destas caixas. Qual é o seu algoritmo para procurá-la?

## Aqui está uma abordagem:



> 1. Monte uma pilha com as caixas que serão analisadas.
> 2. Pegue uma caixa e olhe o que tem dentro dela.
> 3. Se você encontrar outra caixa dentro dela, adicione-a um novo monte para ser verificada mais tarde.
> 4. Se você encontrar uma chave, terminou!
> 5. Repita.

## Aqui está outra abordagem:



> 1. Olhe o que tem dentro da caixa.
> 2. Se encontrar outra caixa, volte ao passo 1.
> 3. Se encontrar a chave, terminou!

# Qual abordagem parece mais fácil?
A primeira abordagem utiliza um *loop* **while**. Enquanto o monte existir, pegue uma caixa e olhe o que tem dentro dela?

```python
def procure_pela_chave(caixa_principal):
    pilha = caixa_principal.crie_uma_pilha_para_busca()

    while pilha:
        caixa = pilha.pop()

        for item in caixa:
            if item.e_uma_caixa():
                pilha.append(item)

            elif item.e_uma_chave():
                print("Achei a chave!")
                return
```

A segunda maneira utiliza a recursão. **Recursão** é quando uma função chama a si mesma. Veja o pseudocógio de como isso funciona.

```python
def procure_pela_chave(caixa):
    for item in caixa:
        if item.e_uma_caixa():
            if procure_pela_chave(item):
                return True

        elif item.e_uma_chave():
            print("Achei a chave!")
            return True

    return False
```

Ambas as abordagens cumprem com a mesma proposta, mas a segunda me parece mais simples. A recursão é usada para tornar a resposta mais clara. Não há nenhum benefício quanto ao desempenho ao utilizar recursão. Na verdade, os loops algumas vezes são melhor para o desempenho do programa.

Muitos algoritmos utilizam a recursão, então é fundamental entender este conceito

# Caso base e caso recursivo

Devido ao fato de a função recursiva chamar a si mesma, é amis fácil escrevê-la erroneamente e acabar em um loop infinito.

É por isso que toda função recursiva tem duas partes: o caso-base e o caso recursivo.

**caso recursivo** é quando a função chama a si mesma.
**caso-base:** é quando a função não chama a si mesma novamente, de forma que o programa não se torna um loop infinito.

```python
def fatorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * fatorial(n - 1)

print(fatorial(5))
```
