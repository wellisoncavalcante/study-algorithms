# Algoritmos gulosos

## O problema do cronograma da sala de aula
Suponha que você tenha uma sala de aula e queira reservar o máximo de aulas possível nela. Assim, recebe-se uma lista das aulas.





O algoritmo é tão simples. Aqui temos a funcionalidade dele:

> 1. Pegue a aula que termina mais cedo. Está é a primeira aula que você colocará nessa sala.
> 2. Agora você precisa pegar uma aula que comece depois da primeira aula. De novo, pegue a aula que termine mais cedo. Esta é a segunda aula que você colocará.

Continue fazendo isso e no final terá sua resposta!



Essa é a beleza dos algoritmos gulosos (também chamados de algoritmos gananciosos): eles são fáceis! Um algoritmo guloso é simples: a cada etapa, deve-se escolher o movimento ideal. Nesse caso, cada vez que você escolhe uma aula, deve escolher a que acaba mais cedo. Em termos técnicos: a cada etapa, escolhe-se a solução ideal, e no fim você tem uma soluçao global ideal. Acredite ou não, esse algoritmo simples acha a solução ideal para esse problema!

Obviamente os algoritmos gulosos nem sempre funcionam, mas eles são tão simples de escrever!

# O problema da mochila.

Suponha que você seja um ladrão ganancioso e esteja em uma loja com sua mochila. Na loja existem diversos itens que você pode roubar. Porém você só pode levar aquilo que caiba na sua mochila, que só suporta 16 quilos.



Você está tentando maximizar o valor dos itens que colocará na sua mochila. Para isso, que algoritmo você usa?

> 1. Pegue o item mais caro que caiba na mochila
> Pegue o próximo item mais caro que caiba na sua mochila, e assim por diante.

Dessa vez, o algoritmo não funciona! Por exemplo, suponha que existam três itens que você possa roubar.



Sua mochila suporta 16 quilos. O aparelho de som é o item mais caro, então você pode roubá-lo. Mas agora não há espaço para mais nada.



Você roubou R$3000 em bens, mas caso tivesse pegado o notebook e o violão, você poderia ter R$3500



Claramente, a estratégia gulosa não oferece a melhor solução aqui, mas fornece um valor bem próximo.

# Exercício 01.
Você trabalha para uma empresa de mobílias e tem de enviar os móveis para todo o país. É necessário encher seu caminhão com caixas, e todas as caixas são de tamanhos diferentes. Você está tentando maximizar o espaço que consegue usar em cada caminhão. Como escolheria as caixas para maximizar o espaço? Proponha uma solução gulosa. Ela lhe dará a solução ideal?

## Minha solução:

> 1. Escolher a maior caixa possível que caiba no caminhão;
> 2. Depois, escolher novamente a outra maior caixa possível que caiba no caminhão;
> 3. Repetir esse processo até que nenhuma outra caixa caiba no caminhão.
> Não. Ela seria uma solução aproximada.

# Exercício 02.
Você está viajando para a Europa e tem sete dias para visitar o maior número de lugares. Para cada lugar que você atribui um valor (o quanto desejar) e estima quanto tempo demora. Como maximizar o total de pontos (passar por todos os lugares que realmente quer ver) durante sua estadia? Proponha uma solução gulosa. Ela lhe dará a solução ideal?

## Minha solução:

> 1. Definir quais lugares eu quero visitar e adicionar o maior número de pontos e estimando quanto tempo visita levará.
> 2. Escolher o primeiro lugar com a maior quantidade de pontos que ainda seja possível visitar dentro do tempo disponível
> 3. Depois, escolher novamente o lugar com a maior quantidade de pontos que caiba no tempo restante.
> 4. Repetir esse processo até complementar o limite máximo de 7 dias.
