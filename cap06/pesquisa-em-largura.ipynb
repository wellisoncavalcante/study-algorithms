# Pesquisa em largura
A pesquisa em largura permite encontrar o menor caminho entre dois objetos. Porém o menor caminho pode significar tantas coisas! Para exemplificar, é possível usar pesquisa em largura para:
> 1. Escrever um algoritmo de inteligência artificial que calcula o menor número de movimentos necessários para a vitória em uma partida de damas.
> 2. Criar um corretor ortográfico (o qual calcula o menor número de edições de edições para transformar a palavra digitada incorretamente em uma palavra real; por exemplo, para modificar LEITOT -> LEITOR é necessário apenas uma edição.)
> 3. Encontrar o médico conveniado ao seu plano de saúde que está mais próximo de você.

## Este algoritmo ajuda a responder a dois tipos de pergunta:
> 1. Existe algum caminho do vértice A até o vértice B?
> 2. Qual o caminho mínimo do vértice A até o vértice B?

## Exercício 01
Encontre o menor caminho do ínicio ao fim.



## Minha solução:
Menor caminho em vermelho.



## Exercício 02
Encontre o menor caminho de "jato" até "gato"



## Minha solução:
Menor caminho em vermelho.
