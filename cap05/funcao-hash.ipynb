# Funções hash
Uma função hash é uma função na qual você insere uma string e depois disso, a função retorna o número.



Você pode pensar que não existe um padrão indicando qual número será rretornado após a inserção de uma string, mas existem alguns requisitos para uma função hash:

> Ela deve ser consistente. Imagine que você insere a string "maçã" e recebe o número 4. Todas as vezes que você inserir "maçã", a função deverá retornar o número "4"; caso contrário, sua tabela hash não funcionará corretamente.

> A função deve mapear diferentes palavras para diferentes números. Desta forma, uma função hash não será útil se ela sempre retornar "1", independente da palavra inserida. No melhor caso, cada palavra diferente deveria ser mapeada e ligada a um número diferente.

Então, uma função hash mapeia strings e as relaciona a números.



Você armazenará os preços das mercadorias neste array. Vamos adicionar o preço de uma maçã. Insira "maçã" na função hash.

Ela retornará o valor "3". Agora, vamos armazenar o preço da maçã no índice 3 do array.



Adicionar "leite" na função hash no índice 0. A função hash retornou "0". Agora, armazene o preço do leite no índice 0.



Continue e, eventualmente, todo o array estará repleto de preços.



Agora, você poderá perguntar "Ei, qual é o preço do abacate?" e não será necessário procurar o preço deste produto no array. Em vez disso, insira "abacate" na função hash.



Ela informará o preço armazenado no índice 4.
