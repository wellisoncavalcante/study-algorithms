# Exercício
Suponha que você tenha estas quatro funções hash que operam com strings.

a. Retorne "1" para qualquer entrada.
b. Use o comprimento da string como índice.
c. Use o primeiro caractere da string como índice. Assim, todas as strings que iniciam com a letra a são hasheadas juntas e assim por diante.
d. Mapeie cada letra para um número primo: a = 2, b = 3, c = 5, d = 7, e = 11, e assim por diante. Para uma string, a função hash é a soma de todos os caracteres-módulo conforme o tamanho da hash. Se o tamanho de sua hash for 10, por exemplo, e a string for "bag", o índice será (3 + 2 + 7) % 10 = 22 % 10 = 2.

Para cada um destes exemplos, qual função hash fornecerá uma boa distribuição? Considere que o tamanho da tabela hash tenha dez espaços.

# Exercício 01.
Uma lista telefônica em que as chaves são os nomes e os valores são os números telefônicos. Os nomes são os seguintes: Esther, Ben, Bob e Dan.

## Minha solução:
**Utilizando a função hash da letra a:**
A função hash da letra a não oferece uma boa distribuição, porque para qualquer entrada será no índice 1. Portanto, causaria colisões entre todos os elementos.

**Utilizando a função hash da letra b:**
A função hash da letra b não oferece uma boa distribuição, visto que causaria colisão entre "Bob", "Ben" e "Dan" porque possuem o mesmo comprimento de string e cairiam no mesmo índice.

**Utilizando a função hash da letra c:**
A função hash da letra c não fornece uma boa distribuição, visto que "Bob" começa com a string "B" e "Ben" começa com a mesma string "B", ocasionando colisão entre os elementos. 

**Utilizando a função hash da letra d:**
A função hash da letra d oferece uma boa distribuição, visto que "Esther" ficaria no índice 0, "Bob" no índice 3,"Ben" no índice 7 e "Dan" no índice 2



# Exercício 02.
Um mapeamento do tamanho de baterias e sua devida potência. Os tamanhos são: A, AA, AAA e AAAA

**Utilizando a função hash da letra a:**
A função hash da letra a não oferece uma boa distribuição, porque para qualquer entrada será no índice 1. Portanto, causaria colisões entre todos os elementos.

**Utilizando a função hash da letra b:**
A função hash da letra b oferece uma boa distribuição, visto que para a string "A" ela estará no índice 1, "AA" no índice 2, "AAA" no índice 3 e por fim "AAAA" no índice 4.

**Utilizando a função hash da letra c:**
A função hash da letra c não fornece uma boa distribuição, visto que todos os elementos começam com a string "A"

**Utilizando a função hash da letra d:**
A função hash da letra d oferece uma boa distribuição, visto que A ficaria no índice 2, AA no índice 4, AAA no índice 6 e AAAA no índice 8


# Exercício 03.
Um mapeamento de títulos de livros e autores. Os títulos são: Maus, Fun Home e Watchmen

**Utilizando a função hash da letra a:**
A função hash da letra a não oferece uma boa distribuição, porque para qualquer entrada será no índice 1. Portanto, causaria colisões entre todos os elementos.
**Utilizando a função hash da letra b:**
A função hash da letra b não oferece uma boa distribuição, porque cada há colisão entre os elementos "Fun Home" e "Watchmen", o elemento "Maus" ficará no índice 4, "Fun Home" no índice 8 e "Watchmen" no índice 8.

**Utilizando a função hash da letra c:**
A função da letra c oferece uma boa distribuição, visto que cada elemento começa com uma caractere diferente, então cada um terá o próprio índice

**Utilizando a função hash da letra d:**
A função da letra d oferece uma boa distribuição, visto que cada elemento ficará em um índice diferente, "Maus" fica no índice 3, "Fun Home" fica no índice 7 e "Watchmen" no índice 5.
